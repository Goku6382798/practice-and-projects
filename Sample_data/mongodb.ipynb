```python
from pymongo import MongoClient, WriteConcern
from urllib.parse import quote_plus
from datetime import datetime
from pymongo.errors import PyMongoError
import pandas as pd
```

```python
    user = "gautam"
    password = quote_plus("Batman")
    uri = f"mongodb+srv://{user}:{password}@testing.rox2wrj.mongodb.net/?retryWrites=true&w=majority&appName=testing"

    client = MongoClient(uri, serverSelectionTimeoutMS=5000)

    try:
        client.admin.command("ping")
        print("Connected")

        print("DBs before:", client.list_database_names())

        db = client.get_database("retail_db", write_concern=WriteConcern(w="majority"))
        coll = db.get_collection("zzz_healthcheck", write_concern=WriteConcern(w="majority"))

        doc = {"_created_at": datetime(), "note": "healthcheck"}
        res = coll.insert_one(doc)
        print("Insert acknowledged:", res.acknowledged, "| inserted_id:", res.inserted_id)

        print("Collections now:", db.list_collection_names())
        print("Doc count:", coll.count_documents({}))

        print("DBs after:", client.list_database_names())

    except PyMongoError as e:
        print("PyMongoError:", e)
    except Exception as e:
        print("Other error:", e)
```
