Two Pointers
When to use(signals)
* Array/string + need pairs / reverse / remove duplicates
* "Sorted array" mentioned
* "Subarray with condition" sometimes (with sliding window)

WHY 

* Maintain two indices to shrink search space in O(n) instead of nested loops.

Note
* The nums must need to be sorted for this technique.

```python
def has_pair_sum(nums, target):
    l, r = 0, len(nums) - 1
    while l < r:
        s = nums[l] + nums[r]
        if s == target:
            return True
        if s < target:
            l += 1
        else:
            r -= 1
    return False
```

```python
has_pair_sum([1,2,3,4,5,6], 9)
```

Sliding Window (Fixed + Variable)
When to use (signals)
* 'subarray/substring'
* "maximum/minimum length"
* "at most/ at least/ exactly k"
* "longest substring without repeating"

WHY
* Window expands and shrinks while maintaining a condition --> 0(n)

Recognition rule:
* If you see "longest/shortest subarray/substring" --> sliding window is usually in the top 2 candidates

Trap: If negative exist and it's about sum >= K, sliding window may fail --> consider prefix sum + hashmap.

```python
def max_sum_k(nums, k):
    window_sum = sum(nums[:k])
    best = window_sum
    for i in range(k, len(nums)):
        window_sum += nums[i] - nums[i-k]
        best = max(best, window_sum)
    return best
```

```python
max_sum_k([2,6,4,8,2,1,9], 3)
```

```python
from collections import defaultdict

def longest_k_distinct(s, k):
    freq = defaultdict(int)
    l = 0
    best = 0

    for r, ch in enumerate(s):
        freq[ch] += 1

        while len(freq) > k:
            left_ch = s[l]
            freq[left_ch] -= 1
            if freq[left_ch] == 0:
                del freq[left_ch]
            l += 1

        best = max(best, r - 1 + 1)
        return best
```
