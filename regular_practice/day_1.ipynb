```python
def binary_search(arr, target):  ##It needs to be in series or set
    l, r = 0, len(arr)-1
    while l <= r:
        mid = (l+r)//2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            l = mid+1
        else:
            r = mid-1
    return -1

print(binary_search([1,3,5,7,9], 7))
print(binary_search([1,3,5,7,9], 4))
```

```python
class Node:
    def __init__(self,val):
        self.val = val
        self.next = None

    def __repr__(self):
        return f"Node({self.val})"
    
class LinkedList:
    def __init__(self):
        self.head = None

    def insert_at_head(self,val):
        new_node = Node(val)
        new_node.next = self.head
        self.head = new_node

    def insert_at_tail(self,val):
        new_node = Node(val)
        if self.head is None:
            self.head = new_node
            return
        curr = self.head
        while curr.next:
            curr = curr.next
        curr.next = new_node

    def from_list(self,values):
        for v in values:
            self.insert_at_tail(v)

    def display(self):
        curr = self.head
        while curr:
            print(curr.val, end=" -> ")
            curr = curr.next
        print("None")

    def reverse_iterative(self):
        prev, curr = None, self.head
        while curr:
            nxt = curr.next
            curr.next = prev
            prev = curr
            curr = nxt
        self.head = prev    

    def reverse_recursive(self):
        def _reverse(node):
            if node is None or node.next is None:
                return node
            new_head = _reverse(node.next)
            node.next.next = node
            node.next = None
            return new_head
        self.head = _reverse(self.head)


ll = LinkedList()

ll.from_list([10,20,30,40,50,60,70])
ll.display()

ll.insert_at_head(5)
ll.display()

ll.insert_at_head(1)
ll.insert_at_tail(50)
ll.display()

ll.reverse_iterative()
ll.display()
```

```python
def is_balanced(expr):
    stack = []
    match = {')':'(',']':'[','}':'{',}

    for ch in expr:
        if ch in "([{":
            stack.append(ch)
        elif ch in ")]}":
            if not stack:
                return False
            top = stack.pop()
            if match[ch] != top:
                return False
    return not stack

print(is_balanced("([)]"))
print(is_balanced("([])"))
```

```python
class MyQueue:
    def __init__(self):
        from collections import deque
        self.q = deque()

    def enqueue(self, val):
        self.q.append(val)

    def dequeue(self):
        if not self.is_empty():
            return self.q.popleft()
        
    def peek(self):
        if not self.is_empty():
            return self.q[0]
        return None
    
    def is_empty(self):
        return len(self.q) == 0

    def __repr__(self):           
        return f"MyQueue({list(self.q)})"    
```

```python
def two_sum(arr, target):
    seen = {}
    for i, num in enumerate(arr):
        complement = target - num
        if complement in seen:
            return (seen[complement], i)
        seen[num] = i
    return None 

two_sum([5,7,6,8,12,65,22,3], 17)
```

```python
def length_of_longest_substring(s: str) -> int:
    last = {}
    l = 0
    best = 0

    for r, ch in enumerate(s):
        if ch in last and last[ch] >= l:
            l = last[ch] + 1
        last[ch] = r
        best = max(best, r - l + 1)

    return best 

print(length_of_longest_substring("aabbccddeee"))
print(length_of_longest_substring("abcdefgh"))
print(length_of_longest_substring("pwpwpwpwpwkkyyyy"))
print(length_of_longest_substring("plplplplpjpjpjsasjkds")) 
```

```python
from collections import Counter

def find_anagrams(s: str, p: str):
    result = []
    p_count = Counter(p)
    window_count = Counter(s[:len(p)])

    if window_count == p_count:
        result.append(0)

    for i in range(len(p), len(s)):
        start_char = s[i-len(p)]
        new_char = s[i]

        window_count[new_char] +=1
        window_count[start_char] -=1
        if window_count[start_char] == 0:
            del window_count[start_char]

        if window_count == p_count:
            result.append(i-len(p)+1)

    return result

print(find_anagrams("cbaebabcdacd","abc"))
print(find_anagrams("ababab","ab"))
```

```python
def longest_subarray_sum_k(arr, K):
    first = {0: -1}
    prefix = 0
    best = 0

    for j, x in enumerate(arr):
        prefix += x

        need = prefix - K
        if need in first:
            best = max(best, j - first[need])

        if prefix not in first:
            first[prefix] = j

    return best
```

```python
def longest_subarray_sum_k_indices(arr, K):
    first = {0: -1}     # prefix 0 @ -1 (empty prefix)
    prefix = 0
    best_len, best_pair = 0, None

    for j, x in enumerate(arr):
        prefix += x
        need = prefix - K

        if need in first:
            curr_len = j - first[need]
            if curr_len > best_len:
                best_len = curr_len
                best_pair = (first[need] + 1, j)  # start..end (inclusive)

        # store only the FIRST time we see this prefix
        if prefix not in first:
            first[prefix] = j

    return best_len, best_pair
```

```python
from collections import defaultdict

def count_subarrays_sum_k(arr, K):
    freq = defaultdict(int)
    freq[0] = 1
    prefix = 0
    count = 0

    for x in arr:
        prefix += x
        need = prefix - K
        count += freq[need]
        freq[prefix] += 1

    return count

print(count_subarrays_sum_k([1,2,3,4,5,6],5))
print(count_subarrays_sum_k([1,2,3,4,5,6],7))
print(count_subarrays_sum_k([1,2,3,4,5,6],11))
```

```python
from collections import deque

def sliding_window_max(arr, k):
    if k <= 0 or not arr:
        return []
    dq = deque()
    out = []

    for i, x in enumerate(arr):
        while dq and dq[0] <= i - k:
            dq.popleft()

        while dq and arr[dq[-1]] <= x:
            dq.pop()

        dq.append(i)

        if i >= k - 1:
            out.append(arr[dq[0]])
    return out

print(sliding_window_max([1,3,-1,-3,5,3,6,7],3))
print(sliding_window_max([9,8,7,6,5],2))
print(sliding_window_max([1,1,1,1],1))
```

```python
from collections import deque

def first_negative_in_window(arr, k):
    dq = deque()
    out = []

    for i, x in enumerate(arr):
        while dq and dq[0] <= i - k:
            dq.popleft()

        if x < 0:
            dq.append(i)

        if i >= k - 1:
            if dq:
                out.append(arr[dq[0]])
            else:
                out.append(0)
    return out
```

```python
from collections import defaultdict

def longest_at_most_distinct(s: str, K: int) -> int:
    if K <= 0:
        return 0
    count = defaultdict(int)
    l = 0
    best = 0
    for r, ch in enumerate(s):
        count[ch] += 1
        while len(count) > K:
            left_ch = s[l]
            count[left_ch] -= 1
            if count[left_ch] == 0:
                del count[left_ch]
            l += 1
        best = max(best, r - 1 + 1)
    return best
```

```python
def longest_exactly_k_distinct(s: str, K: int) -> int:
    if K <= 0:
        return 0
    return longest_exactly_k_distinct(s, K) - longest_at_most_distinct(s, K-1)   

print(longest_at_most_distinct("aabacbebebe", 3))
print(longest_at_most_distinct("eceba", 2))
```

```python
def min_subarray_len_at_least_s(nums, S):
    l = 0
    curr = 0
    ans = float('inf')
    for r, x in enumerate(nums):
        curr += x
        while curr >= S:
            ans = min(ans, r - l + 1)
            curr -= nums[l]
            l += 1
    return 0 if ans == float('inf') else ans

print(min_subarray_len_at_least_s([2,3,1,2,4,3], 7))
```

```python
def num_subarray_product_less_than_k(nums, K):
    if K <= 1: return 0
    prod = 1
    l = 0
    ans = 0
    for r, x in enumerate(nums):
        prod *= x
        while prod >= K:
            prod //= nums[l] if isinstance(prod, int) else prod / nums[l]
            l += 1
        ans += (r - l + 1)
    return ans
print(num_subarray_product_less_than_k([10,5,2,6], 100))
```

```python
from collections import Counter

def min_window_substring(s: str, t: str) -> str:
    if not s or not t:
        return ""
    need = Counter(t)
    need_types = len(need)
    have_types = 0
    window = Counter()
    l = 0
    best_len, best_l, best_r = float('inf'), 0, 0

    for r, ch in enumerate(s):
        window[ch] += 1
        if ch in need and window[ch] == need[ch]:
            have_types += 1

        while have_types == need_types:
            if r - l + 1 < best_len:
                best_len, best_l, best_r = r - l + 1, l, r

            left_ch = s[l]
            window[left_ch] -= 1
            if left_ch in need and window[left_ch] < need[left_ch]:
                have_types -= 1
            l += 1
    return "" if best_len == float('inf') else s[best_l:best_r+1]
```

```python
class Node:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None

def build_sample_tree():
    root = Node(50)
    root.left = Node(30)
    root.right = Node(70)

    root.left.left = Node(20)
    root.left.right = Node(40)
    root.right.left = Node(60)
    root.right.right = Node(80)

    root.left.left.left = Node(10)
    root.left.left.right = Node(25)
    root.left.left.right.left = Node(22)

    root.right.left.left = Node(55)
    root.right.left.right = Node(65)
    root.right.left.right.right = Node(68)
    
    root.right.right.left = Node(75)
    root.right.right.right = Node(90)
    
    return root

def lca(root, p, q):
    if not root or root.val == p or root.val == q:
        return root
    
    left = lca(root.left,p,q)
    right = lca(root.right,p,q)

    if left and right:
        return root
    return left if left else right
```

```python
from collections import deque
class Node:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None

def build_sample_tree():
    root = Node(10)
    root.left = Node(5)
    root.right = Node(15)
    root.left.left = Node(2)
    root.right.right = Node(20)
    return root

def is_balanced(root):
    def check(node):
        if not node:
            return 0
        left = check(node.left)
        if left == -1: return -1
        right = check(node.right)
        if right == -1: return -1

        if abs(left - right) > 1:
            return -1
        return 1 + max(left, right)
    return check(root) != -1
```

```python
from collections import deque
class Node:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None

def build_sample_tree():
    root = Node(10)
    root.left = Node(5)
    root.right = Node(5)
    root.left.left = Node(20)
    root.right.right = Node(20)
    return root

def is_mirror(t1,t2):
    if not t1 and not t2:
        return True
    if not t1 or not t2:
        return False
    return (t1.val == t2.val and
            is_mirror(t1.left, t2.right) and
            is_mirror(t1.right, t2.left))

def is_symmetric(root):
    if not root:
        return True
    return is_mirror(root.left, root.right)
```

```python
from collections import deque
class Node:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None

def build_sample_tree():
    root = Node(5)
    root.left = Node(4)
    root.right = Node(8)
    root.left.left = Node(11)
    root.left.left.left = Node(7)
    root.left.left.right = Node(2)
    root.right.right = Node(13)
    root.right.left = Node(4)
    return root
```

```python
def has_path_sum(root, target):
    if not root:
        return False
    if not root.left and not root.right:
        return target == root.val
    return (has_path_sum(root.left, target - root.val) or
            has_path_sum(root.right, target - root.val))

def all_paths(root):
    res = []
    def dfs(node, path):
        if not node:
            return
        path.append(node.val)
        if not node.left and not node.right:
            res.append(list(path))
        else:
            dfs(node.left, path)
            dfs(node.right, path)
        path.pop()
    dfs(root, [])
    return res

def max_path_sum(root):
    max_sum = float("-inf")
    def dfs(node):
        nonlocal max_sum
        if not node:
            return 0
        left = max(dfs(node.left), 0)
        right = max(dfs(node.right), 0)
        max_sum = max(max_sum, node.val + left + right)
        return node.val + max(left, right)
    dfs(root)
    return max_sum
```

```python
class Node:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None
    def __repr__(self):
        return f"Node({self.val})"
    
def bst_insert(root, key):
    if not root:
        return Node(key)
    if key < root.val:
        root.left = bst_insert(root.left, key)
    elif key > root.val:
        root.right = bst_insert(root.right, key)
    return root

def build_bst(values):
    root = None
    for v in values:
        root = bst_insert(root, v)
    return root

def bst_search_iter(root, key):
    curr = root
    while curr:
        if key == curr.val:
            return curr
        elif key < curr.val:
            curr = curr.left
        else:
            curr = curr.right
    return None

def bst_search_rec(root, key):
    if not root:
        return None
    if key == root.val:
        return root
    if key < root.val:
        return bst_search_rec(root.left, key)
    else:
        return bst_search_rec(root.right, key)
```

```python
class Node:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None

def bst_insert(root, key):
    if not root:
        return Node(key)
    if key < root.val:
        root.left = bst_insert(root.left, key)
    elif key > root.val:
        root.right = bst_insert(root.right, key)
    return root

def inorder(root):
    if not root: return []
    return inorder(root.left) + [root.val] + inorder(root.right)
```

```python
def bst_delete(root, key):
    if not root:
        return None
    
    if key < root.val:
        root.left = bst_delete(root.left, key)
    elif key > root.val:
        root.right = bst_delete(root.right, key)
    else:
        if not root.left and not root.right:
            return None
        if not root.left:
            return root.right
        if not root.right:
            return root.left
        succ = root.right
        while succ.left:
            succ = succ.left
        root.val = succ.val
        root.right = bst_delete(root.right, succ.val)
    return root

def inorder(root):
    if not root: return []
    return inorder(root.left) + [root.val] + inorder(root.right)
```

```python
root = None
for v in [15,10,20,8,12,17,25]:
    root = bst_insert(root, v)

print("Before:", inorder(root))

root = bst_delete(root, 10)
print("after:", inorder(root))


root = bst_delete(root, 20)
print("after:", inorder(root))


root = bst_delete(root, 15)
print("after:", inorder(root))
```

```python
def is_valid_bst(root):
    def dfs(node, lo, hi):
        if not node:
            return True
        if not (lo < node.val < hi):
            return False
        return dfs(node.left, lo, node.val) and dfs(node.right, node.val, hi)
    return dfs(root, float("-inf"), float("inf"))

def kth_smallest(root, k):
    stack, curr = [], root
    while True:
        while curr:
            stack.append(curr)
            curr = stack.left
        curr = stack.pop()
        k -= 1
        if k == 0:
            return curr.val
        curr = curr.right

def bst_floor(root, x):
    res = None
    while root:
        if root.val == x:
            return root.val
        if root.val < x:
            res = root.val
            root = root.right
        else:
            root = root.left
    return res

def bst_ceil(root, x):
    res = None
    while root:
        if root.val == x:
            return root.val
        if root.val > x:
            res = root.val
            root = root.left
        else:
            root = root.right
    return res

def range_sum_bst(root, low, high):
    if not root:
        return 0
    if root.val < low:
        return range_sum_bst(root.right, low, high)
    if root.val > high:
        return range_sum_bst(root.left, low, high)
    return (root.val +
            range_sum_bst(root.left, low, high) + 
            range_sum_bst(root.right, low, high))

def lca_bst(root, p, q):
    curr = root
    while curr:
        if p < curr.val and q < curr.val:
            curr = curr.left
        elif p > curr.val and q > curr.val:
            curr = curr.right
        else:
            return curr
```

```python
root = None
for v in [15,10,20,8,12,17,25]:
    root = bst_insert(root, v)

is_valid_bst(root)
kth_smallest(root, 17)
bst_floor(root, 25)
```
