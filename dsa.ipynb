```python
def prime_number(n):
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return f"{n} is not prime"
    return f"{n} is prime"

prime_number(13)
```

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(6))
```

```python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(6))
```

```python
def reverse_list(lst, index=0):
    if index >= len(lst) // 2:
        return lst
    # Swap the elements
    lst[index], lst[-(index + 1)] = lst[-(index + 1)], lst[index]
    return reverse_list(lst, index + 1)

print(reverse_list([1, 2, 3, 4, 5, 6]))
```

```python
def linear_serch(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1

print(linear_serch([1,2,3,5,6,7,11], 7))
```

```python
def binary_search(arr, target):
    left = 0
    right = len(arr) -1

    while left <= right:
        mid = (left + right) //2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return -1
```

```python
def has_pair(arr, target):
    arr.sort()  # sort first
    left = 0
    right = len(arr) - 1

    while left < right:
        current_sum = arr[left] + arr[right]
        if current_sum == target:
            return True
        elif current_sum < target:
            left += 1
        else:
            right -= 1
    return False
```

```python
print(has_pair([12,13,15,12,16,17,18,11], 24))  # True (12 + 12)
print(has_pair([1, 2, 3, 4], 8))                # False
```

```python
def max_sum_subarray(arr, k):
    window_sum = sum(arr[:k])
    max_sum = window_sum

    for i in range(k, len(arr)):
        window_sum = window_sum - arr[i - k] + arr[i]
        max_sum = max(max_sum, window_sum)

    return max_sum
print(max_sum_subarray([1,2,3,4,5], 4))
```
